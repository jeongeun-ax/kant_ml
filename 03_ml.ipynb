## 선형회귀 실습 

### 학습 데이터 준비 

```python
import pandas as pd
import matplotlib.pyplot as plt

# 선형회귀 모델 임포트
from sklearn.linear_model import LinearRegression

# 한글 폰트
plt.rcParams["font.family"] = "Malgun Gothic"

# 데이터 (딕셔너리)
data = {
    "hours": [1, 2, 3, 4, 5],
    "scores": [47, 54, 65, 73, 87]
}

# 판다스 데이터프레임으로 전환
df = pd.DataFrame(data)
df
```

### 입력 데이터와 정답데이터 분리

```python
# 입력 데이터 (피쳐: Feature) : 2차원 - 독립변수
X =  df[["hours"]]  # 공부시간

# 출력 데이터 (Target) : 1차원 - 정답 : 종속변수 
y = df["scores"]    # 시험점수
```

### 모델 생성 및 학습 

```python
# 선형회귀 모델 객체 생성 
model = LinearRegression()
# 모델 학습 
model.fit(X,y)
```

### 결과 확인 

```python
# 다중 선형회귀의 경우 결과값이 배열로 리터
# 기울기는 배열 리턴 
print(f"기울기: {model.coef_[0]:.2f}")    # 기울기 Weight 
print(f"절편: {model.intercept_:.2f}")    # 절편 Bias
```

### 새로운 데이터로 에측하기 

```python
# 6시간 학습한 학생의 성적 예측 
new_data = pd.DataFrame({"hours": [6]})   # X 독립변수 1개 추가 

predicted_score = model.predict(new_data)
print(f"6시간 학습한 학생의 예상 성적: {predicted_score[0]:.2f}점")
```

### 학습 결과 시각화 

```python
# 학습 데이터에 대한 예측 값
y_pred = model.predict(X)
print("학습 데이터에 대한 예측값:", y_pred)

plt.figure(figsize=(8,5))

# 실제 성적 (산점도)
plt.scatter(X["hours"], y, color="blue", label="실제 성적")

# 예측한 성적 (회귀직선)
plt.plot(X["hours"], y_pred, color="red", label="예측 성적")

# 6시간 학습한 학생의 예측 성적 
plt.scatter(6, predicted_score[0], color="green", label="6시간 예측 성적")

plt.title("공부시간 대비 성적 에측")
plt.xlabel("공부 시간")
plt.ylabel("성적")
plt.legend()
plt.show()
```

### 모델 평가

```python
from sklearn.metrics import mean_absolute_error 
from sklearn.metrics import mean_squared_error  # 오차를 제곱하는 이유: 오차값을 크게 해서 발견에 용이하기 위해 
from sklearn.metrics import r2_score 

# NAE : 오차 절대 평균
# 오차의 크기를 직관적으로 이해 
mae = mean_absolute_error(y, y_pred)
print(f"학습데이터에 대한 평균 절대 오차(MAE): {mae:.2f}")

# MSE : 오차 제곱 평균 
# 제곱하는 이뉴는 큰 오차에 대해서 민감하게 반응 
mse = mean_squared_error(y, y_pred)
print(f"학습데이터에 대한 평균 제곱 오차(MSE): {mse:.2f}")

# R2 Score : 결정계수 
# 모델이 실제 데이터를 얼마만큼 정확한지를 나타내는 지표 
# 0 <==> 1
# 음수일 경우 학습 실패 
r2 = r2_score(y, y_pred)
print(f"학습데이터에 대한 결정 계수(R^2): {r2:.2f}")

```

### 손실 함수 (Loss Function)

- 모델이 예측한 값이 실제 정답과 얼마나 차이가 나는지를 수치로 표현하는 함수 
- MSE 를 대표적으로 사용 

### 학습데이터 준비 

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
# 배치 경사 하강법 (BGD) : 전체 데이터를 사용, 시간, 리소스 많이 사용 
# 확률적 경사 하강법 (SGD) : 확률적으로 1개의 데이터를 추출해서 하강 방법 

from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error

# 학습 데이터 생성 
df = pd.DataFrame({
    "hours": [1, 2, 3, 4, 5],
    "scores": [47, 53, 65, 73, 87]
})

# 독립변수 (입력변수 : Feature)
X = df[["hours"]]
# 종속변수 (정답 : Target)
y = df["scores"]

df
```

```python
# 선형회귀
model = LinearRegression()
# 모델 학습
model.fit(X, y)
# 예측값
y_pred = model.predict(X)
# 결과값
print(f"Weight: {model.coef_[0]:.2f}")
print(f"Bias: {model.intercept_:.2f}")
print(f"MSE: {mean_squared_error(y, y_pred):.4f}")
```

### 확률적 경사하강법 모델 생성
- SGDRegressor 모델 생성
- SGD(Stochastic Gradient Descent)

```python
sgd = SGDRegressor(
    loss="squared_error",       # 제곱 오차 기반 손실함수
    learning_rate="constant",   # 학습률을 일정하게 유지
    eta0=0.01,                  # 초기 학습률
    max_iter=1000,              # 학습횟수 fit() 반복 회수
    shuffle=False,              # 학습 데이터 섞지 않음
    random_state=42             # 난수 시드 설정
)
```

```python
# 전체 데이터를 기준으로 총 몇회 반복
epochs = 1000

# 학습과정 저장 리스트
history = []

for epoch in range(epochs):
    # 전체 데이터 1회 학습 (1 epoch)
    sgd.partial_fit(X, y)

    # 1회 학습 후 예측
    y_pred = sgd.predict(X)

    # 현재 모델의 MSE 계산
    mse = mean_squared_error(y, y_pred)

    # 현재 학습 결과 저장(누적)
    history.append({
        "epoch": epoch + 1,
        "weight": sgd.coef_[0],
        "bias": sgd.intercept_[0],
        "mse": mse
    })

# 학습 기록 => DataFrame 으로 변환
history_df = pd.DataFrame(history)
history_df
```

```python
history_df[ history_df["epoch"].isin([1,5,10,50,100,300,500,1000]) ]
```

### 손실함수 차트

```python
plt.figure(figsize=(10,5))

# Epoch에 따른 MSE 변화
plt.plot(
    history_df["epoch"],
    history_df["mse"],
    color="blue",
    label="학습 손실(Traing Loss)"
)
plt.xlabel("Loss")
plt.ylabel("MSE")
plt.grid(alpha=0.3)
plt.show()
```

### 기울기와 절편 변화 시각화

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 기울기 변화
axes[0].plot(
    history_df["epoch"],
    history_df["weight"],
    color="blue"
)
axes[0].grid(alpha=0.3)
axes[0].set_ylabel("Weight")

# 절편 변화
axes[1].plot(
    history_df["epoch"],
    history_df["bias"],
    color="blue"
)
axes[1].grid(alpha=0.3)
axes[1].set_ylabel("Bias")

plt.show()
```

### 최종 모델로 시험 적적 예측하기

```python
# 최종 학습된 기울기, 절편
print(f"기울기(w): {sgd.coef_[0]:.2f}")
print(f"절편(b): {sgd.intercept_[0]:.2f}") 

# 6시간 학습한 학생
new_data = pd.DataFrame({"hours": [6]})

# 경사하강법으로 학습한 모델로 예측
prediction = sgd.predict(new_data)

print(f"예측 성적: {prediction[0]:.2f}점")
```

## 다중 선형회귀 (Multiple Linear Regression)


### 라이브러리 임포트 및 데이터 준비

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

# 데이터 생성
df = pd.DataFrame({
    "study_hours":    [1, 2, 3, 4, 5, 2, 3, 5, 4, 1],
    "sleep_hours":    [6, 8, 5, 7, 8, 5, 8, 6, 6, 8],
    "practice_count": [10, 15, 20, 25, 30, 12, 28, 22, 32, 8],
    "score":          [47, 58, 60, 76, 87, 49, 73, 77, 79, 52]
})

df 
```

### 상관관계 확인 (차트)

```python
# 1행 3열의 그래프 영역 생성
fig, axes = plt.subplots(1, 3, figsize=(16,5))

# 공부시간 + 성적
sns.scatterplot(
    data=df,
    x="study_hours",
    y="score",
    ax=axes[0]
)

# 수면시간 + 성적
sns.scatterplot(
    data=df,
    x="sleep_hours",
    y="score",
    ax=axes[1]
)

# 문제풀이 개수 + 성적
sns.scatterplot(
    data=df,
    x="practice_count",
    y="score",
    ax=axes[2]
)

plt.show()
```

### 상관 계수 확인

```python
# 모든 컬럼 사이의 상관계수 확인
corr = df.corr()

plt.figure(figsize=(7,5))

sns.heatmap(
    corr,
    annot=True,     # 상관계수 표시
    vmin=-1,        # 최소값
    vmax=1,         # 최대값
    fmt=".2f"       # 소숫점 두자리
)

plt.show()
```

### 입력 데이터(X)와 정답데이터(Y) 분리 

```python
# 복수개의 입력 데이터
features = [
    "study_hours", 
    "sleep_hours",
    "practice_count"
]

# 입력 데이터 (2차원)
X = df[features]

# 정답 데이터 (1차원)
y = df["score"]

print(X.shape)
print(y.shape)
```

### 다중 선형 회귀 모델 생성, 학습 

```python
# 다중 선형회구 모델 객체 생성 
model = LinearRegression()

# 입력 데이터 3, 출력 데이터 1 사용해서 학습
model.fit(X,y)
```

### 학습된 가중치(weight) 와 절편(bias) 확인
- coef_  1개 결과
- coef_  3개 결과

```python
print("가중치", model.coef_)
print("절편", model.intercept_)
```

### 가중치 시각화 

```python
# 입력변수(독립변수, Feature)와 가중치(기울기: weight) DataFrame으로 생성
coef_df = pd.DataFrame({
    "feature": features,
    "weight": model.coef_
})

# coef_df


plt.figure(figsize=(6,2))

# 가로바 막대 그래프
plt.barh(
    coef_df["feature"], # 세로축에 표시 
    coef_df["weight"], 
    color = "blue"
)

plt.show()
```

### 새로운 학생의 데이터로 예측
- 공부시간: 4시간
- 수면시간: 7시간
- 문제풀이: 30개 

```python
# 새로운 데이터 생성
new_data = pd.DataFrame({
    "study_hours":[4],
    "sleep_hours": [7],
    "practice_count": [30]
})

# 학습한 모델로 시험 성적 예측 
prediction = model.predict(new_data)

# 예측 결과
print(f"예측 성적: {prediction[0]:.2f}점")
```

```python
# 기울기: 5.80222725 2.68100001 0.63219852
# 기울기: 17.66

# 수동으로 계산 해보기 
# (4 * 5.80222725) + (7 * 2.68100001) + (30 * 0.63219852) +17.66

x1 = 4
x2 = 7 
x3 = 30

# 학습된 가중치(기울기)
w1, w2, w3 = model.coef_
# 학습된 절편(편향, bias)
b = model.intercept_

# 회귀공식에 대입 
result = w1 * x1 + w2 * x2 + w3 * x3 + b

print(f"직접 계산: {result:.4f}")
print(f"predict 계산: {prediction[0]:.4f}")
```

### 실제 성적과 예측 성적의 오차 비교 

```python
# 전체 학습 데이터에 대한 예측값
y_pred = model.predict(X)

# 실제값과 예측값을 비교하기 위해 복사본 생성 
result_df = df.copy()

# 없는 열을 명시하면 자동으로 추가된다
result_df["예측값"] = y_pred
# 오차 계산해서 새로운 열로 추가 
result_df["예측오차"] = y - y_pred

result_df.round(2)
```

### 모델 평가 

```python
# 오차 절대 평균 (MAE)
mae = mean_absolute_error(y, y_pred)
# 오차 제곱 평균 (MSE)
mse = mean_squared_error(y, y_pred)
# R2 Score (결정계수)
r2 = r2_score(y, y_pred)

print(f"MAE: {mae:.4f}")
print(f"MSE: {mse:.4f}")
print(f"R2: {r2:.4f}")
```

### 학습 데이터와 테스트 데이터 분리 

```python
# 학습 데이터와 테스트 데이터 분리
# X_train, X_test, y_train, y_test
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y,
    test_size=0.3,
    random_state=42
)

print("X_train", X_train.shape)
print("X_test", X_test.shape)
print("y_train", y_train.shape)
print("y_test", y_test.shape)
```

### 모델 학습 

```python
test_model = LinearRegression()
# 70% 데이터로 학습 
test_model.fit(X_train, y_train)
# 학습하지 않은 데이터로 결과 예측
test_pred = test_model.predict(X_test)

print(f"MAE: {mean_absolute_error(y_test, test_pred):.4f}")
print(f"MSE: {mean_squared_error(y_test, test_pred):.4f}")
print(f"R2: {r2_score(y_test, test_pred):.4f}")
```

### 경사하강법을 이용한 다중 선형회귀
- SGDRegressor 사용 
- 입력변수 3개, 출력변수 1개


```python
from sklearn.linear_model import SGDRegressor


# 표준화된 학습 데이터 (70%)로 모델 학습 
sgd.fit(X_train_scaled, y_train)

# 표준화된 검증 데이터(30%)로 모델 예측 
sgd_pred = sgd.predict(X_test_scaled)

# 모델 평가
print(f"MAE: {mean_absolute_error(y_test, sgd_pred):.4f}")
print(f"MSE: {mean_squared_error(y_test, sgd_pred):.4f}")
print(f"R2: {r2_score(y_test, sgd_pred):.4f}")
```

```python
sgd = SGDRegressor(
    loss="squared_error",       #MSE 손실함수
    learning_rate="constant",   #일정하게 학습
    eta0=0.01,                  # 초기 학습률
    random_state=42             # 난수 발생 seed
)

history = []

# 500 Epoch 학습 
for epoch in range(500):
    # 1 Epoch 한번 학습
    sgd.partial_fit(X_train_scaled, y_train)
    pred = sgd.predict(X_train_scaled)

    history.append({
        "epoch": epoch + 1, 
        "w1": sgd.coef_[0], 
        "w2": sgd.coef_[1], 
        "w3": sgd.coef_[2], 
        "bias": sgd.intercept_[0],
        "mse": mean_squared_error(y_train, pred)
    })

history_df = pd.DataFrame(history)
history_df.tail()
```

```python
# 손실곡선
# X = epoch
# y = mse
# plot

plt.plot(
    history_df["epoch"],
    history_df["mse"]
)

plt.show()

# w1 가중치 차트 
# X = epoch 

plt.plot (
    history_df["epoch"],
    history_df["w1"]
)

plt.show()
```
